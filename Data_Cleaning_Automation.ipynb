```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\LOQ\Documents\data-cleaning-automation-project\dataset\sales_data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)

print("Duplicates Removed")
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)

print("Missing Values Handled")
```

```python
df.columns = df.columns.str.strip().str.lower()

print(df.columns)
```

```python
summary = df.describe()

print(summary)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['total_bill'], kde=True)

plt.title("Distribution of Total Bills")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['tip'])

plt.title("Tip Distribution")

plt.show()
```

```python
df.to_csv(
    r"C:\Users\LOQ\Documents\data-cleaning-automation-project\output\cleaned_sales_data.csv",
    index=False
)

print("Cleaned Dataset Exported")
```
